```python
import pandas as pd
import numpy as np
import os

filename = next(iter(uploaded))

df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 records:")
display(df.head())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np

# Check dataset
print("Original shape:", df.shape)

# Remove duplicate records
df = df.drop_duplicates()

# Handle missing values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

df["local_authority_name"] = (
    df["local_authority_name"].fillna("Unknown")
)

# Sort records by authority and year
df = df.sort_values(
    ["local_authority_name", "year"]
).reset_index(drop=True)

print("Prepared shape:", df.shape)
print(df.head())
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["local_authority_name"],
    dtype=int
)

print("Encoded shape:", df_encoded.shape)
print(df_encoded.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

TARGET = "all_motor_vehicles"

# Keep the latest year for testing
years = sorted(df_encoded["year"].unique())

if len(years) < 3:
    raise ValueError(
        "At least three distinct years are needed "
        "for a chronological train/validation/test split."
    )

train_years = years[:-2]
val_year = years[-2]
test_year = years[-1]

train_mask = df_encoded["year"].isin(train_years)
val_mask = df_encoded["year"] == val_year
test_mask = df_encoded["year"] == test_year

# Exclude target and ID from the features
excluded = [
    TARGET,
    "local_authority_id"
]

feature_cols = [
    c for c in df_encoded.columns
    if c not in excluded
]

# Convert values to numeric arrays
X_raw = df_encoded[feature_cols].astype(float).values
y_raw = df_encoded[[TARGET]].astype(float).values

x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()

x_scaler.fit(X_raw[train_mask])
y_scaler.fit(y_raw[train_mask])

X = x_scaler.transform(X_raw)
y = y_scaler.transform(y_raw)

print("Training years:", train_years)
print("Validation year:", val_year)
print("Test year:", test_year)
```

```python
TIME_STEPS = 3

X_seq, y_seq, seq_years = [], [], []

for authority in df["local_authority_name"].unique():
    idx = df.index[
        df["local_authority_name"] == authority
    ].to_numpy()

    idx = idx[
        np.argsort(df.loc[idx, "year"].to_numpy())
    ]

    for j in range(TIME_STEPS, len(idx)):
        prev_idx = idx[j-TIME_STEPS:j]
        target_idx = idx[j]

        X_seq.append(X[prev_idx])
        y_seq.append(y[target_idx])
        seq_years.append(df.loc[target_idx, "year"])

X_seq = np.array(X_seq)
y_seq = np.array(y_seq)
seq_years = np.array(seq_years)

X_train = X_seq[np.isin(seq_years, train_years)]
y_train = y_seq[np.isin(seq_years, train_years)]

X_val = X_seq[seq_years == val_year]
y_val = y_seq[seq_years == val_year]

X_test = X_seq[seq_years == test_year]
y_test = y_seq[seq_years == test_year]

print("Training sequences:", X_train.shape)
print("Validation sequences:", X_val.shape)
print("Testing sequences:", X_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

tf.random.set_seed(42)

model = Sequential([
    tf.keras.Input(shape=(3, 213)),
    LSTM(32, return_sequences=False),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=30,
    batch_size=32,
    callbacks=[early_stop],
    shuffle=False,
    verbose=1
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

pred_scaled = model.predict(X_test)

actual = y_scaler.inverse_transform(
    y_test.reshape(-1, 1)
).ravel()

predicted = y_scaler.inverse_transform(
    pred_scaled
).ravel()

mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))

print(f"MAE: {mae:,.2f}")
print(f"RMSE: {rmse:,.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(actual, label="Actual", marker="o")
plt.plot(predicted, label="Predicted", marker="x")

plt.title("Actual vs Predicted Vehicle Volume (2019)")
plt.xlabel("Test observation")
plt.ylabel("All motor vehicles")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Predictions
train_pred_scaled = model.predict(X_train, verbose=0)
test_pred_scaled = model.predict(X_test, verbose=0)

# Convert predictions back to original units
train_actual = y_scaler.inverse_transform(
    y_train.reshape(-1, 1)
).ravel()

train_pred = y_scaler.inverse_transform(
    train_pred_scaled
).ravel()

test_actual = y_scaler.inverse_transform(
    y_test.reshape(-1, 1)
).ravel()

test_pred = y_scaler.inverse_transform(
    test_pred_scaled
).ravel()

# Calculate metrics
train_mae = mean_absolute_error(train_actual, train_pred)
train_rmse = np.sqrt(
    mean_squared_error(train_actual, train_pred)
)

test_mae = mean_absolute_error(test_actual, test_pred)
test_rmse = np.sqrt(
    mean_squared_error(test_actual, test_pred)
)

# Optional approximate accuracy based on mean absolute percentage error
def approximate_accuracy(actual, predicted):
    mask = actual != 0
    if not np.any(mask):
        return np.nan
    mape = np.mean(
        np.abs(
            (actual[mask] - predicted[mask]) / actual[mask]
        )
    ) * 100
    return max(0, 100 - mape)

train_accuracy = approximate_accuracy(train_actual, train_pred)
test_accuracy = approximate_accuracy(test_actual, test_pred)

print("===== TRAINING PERFORMANCE =====")
print(f"Training MAE: {train_mae:,.2f}")
print(f"Training RMSE: {train_rmse:,.2f}")
print(f"Approximate training accuracy: {train_accuracy:.2f}%")

print("\n===== UNSEEN TEST PERFORMANCE =====")
print(f"Test MAE: {test_mae:,.2f}")
print(f"Test RMSE: {test_rmse:,.2f}")
print(f"Approximate test accuracy: {test_accuracy:.2f}%")
```
